# Индивидуальное задание. Набор данных Penguins

```python
import numpy as np
import matplotlib.pyplot as plt

# Явно задаём типы столбцов: измерения - float, категории - строки
dtype = [('id', 'i4'), ('species', 'U9'), ('island', 'U9'),
         ('bill_length_mm', 'f8'), ('bill_depth_mm', 'f8'),
         ('flipper_length_mm', 'f8'), ('body_mass_g', 'f8'),
         ('sex', 'U7'), ('year', 'i4')]

# Пустые значения в CSV: числовые -> nan, пол -> пустая строка
data = np.genfromtxt('datasets/penguins/penguins.csv', delimiter=',', names=True,
                     dtype=dtype, missing_values='', filling_values={'sex': ''})

print(data.shape)
print(data[:5])
```

```python
print(type(data))
for name in data.dtype.names:
    print(name, data.dtype[name])
```

```python
print('Наблюдений:', data.shape[0], '| Полей:', len(data.dtype.names))

for name in ['bill_length_mm', 'bill_depth_mm', 'flipper_length_mm', 'body_mass_g']:
    col = data[name]
    print(f'{name}: пропусков {np.isnan(col).sum()}, среднее {np.nanmean(col):.2f}, '
          f'min {np.nanmin(col)}, max {np.nanmax(col)}')

for name in ['species', 'island', 'sex', 'year']:
    values, counts = np.unique(data[name], return_counts=True)
    print(name, dict(zip(values.tolist(), counts.tolist())))
```

```python
fields = ['bill_length_mm', 'bill_depth_mm', 'flipper_length_mm', 'body_mass_g']
labels = ['Bill Length', 'Bill Depth', 'Flipper Length', 'Body Mass']
species_order = ['Adelie', 'Chinstrap', 'Gentoo']
styles = ['ro', 'g^', 'bs']

# Строим все шесть двумерных проекций
figure = 1
for i in range(len(fields)):
    for j in range(i + 1, len(fields)):
        plt.figure(figure)
        for species, style in zip(species_order, styles):
            mask = data['species'] == species
            x, y = data[fields[i]][mask], data[fields[j]][mask]
            keep = ~np.isnan(x) & ~np.isnan(y)
            plt.plot(x[keep], y[keep], style, label=species)
        plt.legend(bbox_to_anchor=(1.05, 1), loc=2, borderaxespad=0.)
        plt.xlabel(labels[i])
        plt.ylabel(labels[j])
        figure += 1
plt.show()
```

```python
# Распределения числовых признаков
fig, axes = plt.subplots(2, 2, figsize=(10, 7))
for ax, name, label in zip(axes.ravel(), fields, labels):
    values = data[name][~np.isnan(data[name])]
    ax.hist(values, bins=20)
    ax.set_title(label)
    ax.set_xlabel(label)
    ax.set_ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
# Разброс признаков по видам
fig, axes = plt.subplots(2, 2, figsize=(10, 7))
for ax, name, label in zip(axes.ravel(), fields, labels):
    groups = [data[name][(data['species'] == species) & ~np.isnan(data[name])]
              for species in species_order]
    ax.boxplot(groups, tick_labels=species_order)
    ax.set_title(label)
    ax.set_ylabel(label)
plt.tight_layout()
plt.show()
```
